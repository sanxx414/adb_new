```python
from pyspark.sql.functions import col, lit, sum, avg, count, max, min, countDistinct, approx_count_distinct

customer_df=spark.read.format("json").load("/Volumes/adf_sample/default/adf_sample/customer.json")
display(customer_df)
```

```python
customer_df2=customer_df.select("*")
display(customer_df2)
```

```python
from pyspark.sql.functions import col
customer_df2=customer_df.select(
    col("salutation"),
    col("firstname"),
    col("lastname"),
    col("gender"),
    col("email_address"),
    col("birth_country"),
    col("birthdate"),
    col("customer_id"),
    col("address_id"),
    col("is_preffered_customer"),
    col("demographics.buy_potential").alias("buy_potential"),
    col("demographics.credit_rating").alias("credit_rating"),
    col("demographics.education_status").alias("education_status"),
    col("demographics.income_range").alias("income_range"),
    col("demographics.purchase_estimate").alias("purchase_estimate"),
    col("demographics.vehicle_count").alias("vehicle_count")
)
clean_df = customer_df2.na.drop()
display(clean_df)
```

```python
customer_df3=customer_df2.groupBy(col("gender")).count()
display(customer_df3)
```

```python
customer_df4 = customer_df2.withColumnsRenamed({
    "firstname": "f_name",
    "lastname": "l_name",
    "gender": "gender",
    "email_address": "email",
    "birth_country": "b_country",
    "birthdate": "dob",
    "customer_id": "customer",
    "address_id": "address",
    "is_preffered_customer": "preffered_customer"
})
display(customer_df4)
```

```python
customer_df5 = customer_df4.filter(
    (col("credit_rating") == "High Risk") & 
    (col("gender") == "F")
    )
display(customer_df5)
```

```python
customer_df6 = customer_df4.filter(
    (col("dob") >= lit("1980-01-01")) &
    (col("dob") <= lit("1980-12-15"))
 )
display(customer_df6)
```

```python
from pyspark.sql.functions import col, current_date, datediff, floor
customer_df_with_age = customer_df4.withColumn("age", floor(datediff(current_date(), col("dob")) / 365.25))
display(customer_df_with_age)
```

```python
customer_age=customer_df_with_age.groupBy(col("age")).count()
display(customer_age)
```
