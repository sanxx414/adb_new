```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

address_df=spark.read.format("parquet").load("/Volumes/adf_sample/default/adf_sample/address.parquet")
display(address_df)
```

```python
address1_df=address_df.select(
    col("address_id"),
    col("country"),
    col("state"),
    col("city"),
    col("zip"),
    col("street_name"),
    col("street_number"),
    col("location_type")
)
address_final_df = address1_df.na.drop()
display(address_final_df)
```

```python
address_df2=address_final_df.filter(
    (col("state") == "NE")&
    (col("zip") == 67683)
    )
display(address_df2)
```

```python
address3_df=address_final_df.groupBy(col("state")).count().orderBy(col("count").desc())
display(address3_df)
```
